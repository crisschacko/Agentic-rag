```python
def agentic_rag(question, max_attempts=3):

    for attempt in range(1, max_attempts + 1):

        print(f"\n--- Retrieval Attempt {attempt} ---")

        # 1. Retrieve information
        results = retriever.invoke(question)

        context = "\n\n".join(
            [result.page_content for result in results]
        )

        # 2. Verification agent
        verification_prompt = f"""
You are an evidence verification agent.

Question:
{question}

Retrieved information:
{context}

Decide whether the retrieved information is sufficient
to answer the question accurately.

Reply with exactly one word:
SUFFICIENT
or
INSUFFICIENT
"""

        verification = client.responses.create(
            model="gpt-5.6-luna",
            input=verification_prompt
        )

        decision = verification.output_text.strip().upper()

        print("Agent decision:", decision)

        # 3. Generate answer if evidence is sufficient
        if "SUFFICIENT" in decision:

            answer_prompt = f"""
Answer the question using ONLY the retrieved information.

Retrieved information:
{context}

Question:
{question}

Give a clear and concise answer.
"""

            answer = client.responses.create(
                model="gpt-5.6-luna",
                input=answer_prompt
            )

            return answer.output_text

        print("Evidence insufficient. Retrieving again...")

    return "I could not find enough reliable information."


# Test Agentic RAG
question = "What is Agentic RAG?"

answer = agentic_rag(question)

print("\nFINAL ANSWER:")
print(answer)
```

```python
def agentic_rag_with_sources(question, max_attempts=3):

    for attempt in range(1, max_attempts + 1):

        print(f"\n--- Attempt {attempt} ---")

        results = retriever.invoke(question)

        context_parts = []

        for i, result in enumerate(results):
            context_parts.append(
                f"[Source {i+1}]\n{result.page_content}"
            )

        context = "\n\n".join(context_parts)

        verification_prompt = f"""
You are an evidence verification agent.

Question:
{question}

Retrieved sources:
{context}

Is the retrieved information sufficient to answer
the question accurately?

Reply with exactly:
SUFFICIENT
or
INSUFFICIENT
"""

        verification = client.responses.create(
            model="gpt-5.6-luna",
            input=verification_prompt
        )

        decision = verification.output_text.strip().upper()

        print("Decision:", decision)

        if "SUFFICIENT" in decision:

            answer_prompt = f"""
Answer the question using ONLY the retrieved sources.

Question:
{question}

Retrieved sources:
{context}

Give a concise answer.

At the end, write:
Sources used: [Source X], [Source Y]
"""

            answer = client.responses.create(
                model="gpt-5.6-luna",
                input=answer_prompt
            )

            print("\nFINAL ANSWER:")
            print(answer.output_text)

            print("\nRETRIEVED SOURCES:")
            for i, result in enumerate(results):
                print(f"\n[Source {i+1}]")
                print(result.page_content)

            return answer.output_text

        print("Insufficient evidence → retrieving again...")

    return "Unable to find sufficient evidence."


agentic_rag_with_sources(
    "What are the main stages of a RAG system?"
)
```

```python
def agentic_rag_v2(question, max_attempts=3):

    current_query = question

    for attempt in range(1, max_attempts + 1):

        print(f"\n--- Attempt {attempt} ---")
        print("Search query:", current_query)

        # 1. Retrieve
        results = retriever.invoke(current_query)

        context = "\n\n".join(
            [result.page_content for result in results]
        )

        # 2. Verify evidence
        verification_prompt = f"""
You are an evidence verification agent.

Original question:
{question}

Search query:
{current_query}

Retrieved information:
{context}

Is the retrieved information sufficient to answer the
original question accurately?

Reply with exactly:
SUFFICIENT
or
INSUFFICIENT
"""

        verification = client.responses.create(
            model="gpt-5.6-luna",
            input=verification_prompt
        )

        decision = verification.output_text.strip().upper()

        print("Decision:", decision)

        # 3. If sufficient, generate answer
        if "SUFFICIENT" in decision:

            answer_prompt = f"""
Answer the original question using ONLY the retrieved information.

Question:
{question}

Retrieved information:
{context}

Give a concise answer.
"""

            answer = client.responses.create(
                model="gpt-5.6-luna",
                input=answer_prompt
            )

            return answer.output_text

        # 4. Rewrite query for another retrieval attempt
        rewrite_prompt = f"""
Rewrite the search query to retrieve better information.

Original question:
{question}

Previous search query:
{current_query}

Retrieved information:
{context}

Return ONLY the improved search query.
"""

        rewritten = client.responses.create(
            model="gpt-5.6-luna",
            input=rewrite_prompt
        )

        current_query = rewritten.output_text.strip()

        print("Rewritten query:", current_query)

    return "Unable to find sufficient evidence."


# Test
answer = agentic_rag_v2(
    "What is Agentic RAG?"
)

print("\nFINAL ANSWER:")
print(answer)
```

```python
def retrieve_again(question):

    new_query_prompt = f"""
Create a better search query for this question.

Question:
{question}

Return only the improved search query.
"""

    response = client.responses.create(
        model="gpt-5.6-luna",
        input=new_query_prompt
    )

    new_query = response.output_text.strip()

    print("New search query:", new_query)

    new_results = retriever.invoke(new_query)

    return new_results
```

```python
def agentic_rag_final(question):

    # Retrieve
    results = retriever.invoke(question)

    context = "\n\n".join(
        result.page_content for result in results
    )

    # Verify
    decision = verify_evidence(question, context)

    print("Verification:", decision)

    # If sufficient, answer
    if decision == "SUFFICIENT":

        prompt = f"""
Answer the question using ONLY this information:

{context}

Question: {question}
"""

        response = client.responses.create(
            model="gpt-5.6-luna",
            input=prompt
        )

        return response.output_text

    # If insufficient, retrieve again
    print("Evidence insufficient. Retrieving again...")

    new_results = retrieve_again(question)

    new_context = "\n\n".join(
        result.page_content for result in new_results
    )

    prompt = f"""
Answer the question using ONLY this information:

{new_context}

Question: {question}
"""

    response = client.responses.create(
        model="gpt-5.6-luna",
        input=prompt
    )

    return response.output_text
```

```python
import faiss
import numpy as np

embedding_matrix = np.array(embeddings).astype("float32")

dimension = embedding_matrix.shape[1]

index = faiss.IndexFlatL2(dimension)
index.add(embedding_matrix)

print("FAISS index created successfully!")
print("Number of vectors:", index.ntotal)
print("Vector dimension:", dimension)
```

```python
def retrieve_documents(question, k=3):
    # Convert question into an embedding
    question_embedding = model.encode([question]).astype("float32")

    # Search FAISS
    distances, indices = index.search(question_embedding, k)

    results = []

    for i, distance in zip(indices[0], distances[0]):
        results.append({
            "text": chunks[i],
            "distance": float(distance)
        })

    return results
```

```python
def verify_evidence(question, context):

    prompt = f"""
You are an evidence verification agent.

Question:
{question}

Retrieved research:
{context}

Determine whether the retrieved research contains enough
information to answer the question accurately.

Reply with exactly one word:

SUFFICIENT

or

INSUFFICIENT
"""

    result = qwen_generate(
        prompt,
        max_new_tokens=20
    )

    result = result.upper()

    if "SUFFICIENT" in result:
        return "SUFFICIENT"

    return "INSUFFICIENT"
```

```python
question = "What is Retrieval-Augmented Generation?"

results = retrieve_research(question)

context = "\n\n".join(results)

print(verify_evidence(question, context))
```

```python
def agentic_research_rag(question):

    print("QUESTION:")
    print(question)

    # First retrieval
    results = retrieve_research(question)

    context = "\n\n".join(results)

    print("\nInitial verification...")

    decision = verify_evidence(
        question,
        context
    )

    print("Verification:", decision)

    # Re-retrieval if evidence is insufficient
    if decision == "INSUFFICIENT":

        print("\nEvidence insufficient.")
        print("Performing re-retrieval...")

        results = retrieve_research(
            question,
            k=6
        )

        context = "\n\n".join(results)

        decision = verify_evidence(
            question,
            context
        )

        print("Second verification:", decision)

    # Final answer
    if decision == "SUFFICIENT":

        prompt = f"""
Answer the question using ONLY the research evidence below.

Question:
{question}

Research evidence:
{context}

Give a concise, factual answer.
Do not invent information.
"""

        answer = qwen_generate(
            prompt,
            max_new_tokens=250
        )

        return answer

    return "The available research evidence is insufficient to answer reliably."
```
